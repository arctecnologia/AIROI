```python
import ipywidgets as widgets
from IPython.display import display, clear_output
import matplotlib.pyplot as plt
import numpy as np

# 1. Legenda Executiva (Adaptável ao Tema)
titulo = widgets.HTML(value="<h2 style='margin-bottom: 0;'>Dashboard de Viabilidade Econômica (Agile AI)</h2>")
descricao = widgets.HTML(value="""
<div style='background: rgba(128, 128, 128, 0.1); padding: 15px; border-left: 4px solid #2980b9; margin-bottom: 25px; border-radius: 4px;'>
    <p style='font-size: 14px; margin: 0; line-height: 1.5;'>
        <b>Lógica do Modelo:</b> O setup (CAPEX) considera um <b>Squad Ágil Integrado</b> operando com esteiras de CI/CD e DevSecOps. O custo da equipe já absorve os papéis de PO, Scrum Master, Arquitetura, Segurança e Desenvolvimento, garantindo entregas versionadas, governadas e monitoradas.
    </p>
</div>
""")

# 2. Configuração de Layout Refinada
style = {'description_width': '320px'}
input_layout = widgets.Layout(width='95%', margin='0 0 15px 0')
col_layout = widgets.Layout(width='50%', padding='0 15px 0 0')

# 3. Campos de Entrada
html_identificacao = widgets.HTML(value="<h4 style='margin: 10px 0 15px 0; border-bottom: 1px solid rgba(128, 128, 128, 0.3); padding-bottom: 5px;'>0. Identificação da Iniciativa</h4>")
nome_processo = widgets.Text(value='Automação de Conciliação Financeira', description='Nome do Processo:', style=style, layout=input_layout)

html_asis = widgets.HTML(value="<h4 style='margin: 25px 0 15px 0; border-bottom: 1px solid rgba(128, 128, 128, 0.3); padding-bottom: 5px;'>1. Cenário Atual (AS IS - Esforço Manual)</h4>")
horas_manuais = widgets.FloatText(value=320.0, description='Horas Manuais/Mês:', style=style, layout=input_layout)
custo_hora_humano = widgets.FloatText(value=55.0, description='Custo Hora Colaborador (R$):', style=style, layout=input_layout)

html_setup = widgets.HTML(value="<h4 style='margin: 10px 0 15px 0; border-bottom: 1px solid rgba(128, 128, 128, 0.3); padding-bottom: 5px;'>2. Investimento em Engenharia de IA (CAPEX)</h4>")
# Correção do rótulo (description) do campo de entrada
custo_squad = widgets.FloatText(value=135000.0, description='Custos Totais do Squad Integrado (R$):', style=style, layout=input_layout)

html_tobe = widgets.HTML(value="<h4 style='margin: 25px 0 15px 0; border-bottom: 1px solid rgba(128, 128, 128, 0.3); padding-bottom: 5px;'>3. Operação da IA (TO BE) e Ganhos</h4>")
custo_nuvem_tokens = widgets.FloatText(value=4500.0, description='Consumo Mensal Nuvem/Tokens (R$):', style=style, layout=input_layout)
custo_mlops = widgets.FloatText(value=2500.0, description='Sustentação/MLOps Mensal (R$):', style=style, layout=input_layout)
ganho_estrategico = widgets.FloatText(value=12000.0, description='Ganho Estratégico/Receita Mensal (R$):', style=style, layout=input_layout)

botao_calcular = widgets.Button(
    description='Gerar Relatório Executivo',
    button_style='primary',
    layout=widgets.Layout(width='300px', margin='30px 0 30px 0', height='40px')
)

resultado = widgets.Output()

def calcular_e_plotar(b):
    with resultado:
        clear_output(wait=True)

        processo_str = nome_processo.value if nome_processo.value.strip() != "" else "Processo Não Identificado"

        # Matemática Financeira
        custo_mensal_asis = horas_manuais.value * custo_hora_humano.value
        custo_anual_asis = custo_mensal_asis * 12

        investimento_total = custo_squad.value

        custo_mensal_tobe = custo_nuvem_tokens.value + custo_mlops.value
        custo_anual_tobe = custo_mensal_tobe * 12

        economia_operacional_mensal = custo_mensal_asis - custo_mensal_tobe
        ganho_total_mensal = economia_operacional_mensal + ganho_estrategico.value
        ganho_total_anual = ganho_total_mensal * 12

        # Relatório de Texto Executivo
        print(f"📊 RESUMO EXECUTIVO: {processo_str.upper()}")
        print("=" * 85)
        print(f"💰 Custos Totais do Squad Integrado:            R$ {investimento_total:,.2f}".replace(",", "X").replace(".", ",").replace("X", "."))
        print(f"🔻 Custo Anual AS IS (Manual):                  R$ {custo_anual_asis:,.2f}".replace(",", "X").replace(".", ",").replace("X", "."))
        print(f"🤖 Custo Anual TO BE (IA e Sustentação):        R$ {custo_anual_tobe:,.2f}".replace(",", "X").replace(".", ",").replace("X", "."))
        print(f"📈 Ganho Estratégico Anual (Receita/Valor):     R$ {(ganho_estrategico.value * 12):,.2f}".replace(",", "X").replace(".", ",").replace("X", "."))
        print("-" * 85)

        if ganho_total_mensal <= 0:
            print("🔴 Alerta: O projeto não gera fluxo de caixa positivo. Revise os custos operacionais ou do Squad.")
            return

        roi = ((ganho_total_anual - investimento_total) / investimento_total) * 100
        payback = investimento_total / ganho_total_mensal

        print(f"✅ Benefício Líquido Anual:                     R$ {ganho_total_anual:,.2f}".replace(",", "X").replace(".", ",").replace("X", "."))
        print(f"🚀 ROI Estimado (1 Ano):                        {roi:.2f}%")
        print(f"⏱️ Payback (Retorno do Capital):                {payback:.1f} meses")

        # Geração de Gráficos (Acessibilidade e Alto Contraste)
        plt.style.use('default')
        fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 5))
        fig.patch.set_facecolor('#ffffff')

        labels = ['Custo AS IS\n(Manual)', 'Custo TO BE\n(IA)']
        valores = [custo_anual_asis, custo_anual_tobe]
        cores = ['#e74c3c', '#2ecc71']

        ax1.set_facecolor('#f8f9fa')
        ax1.bar(labels, valores, color=cores, width=0.5)
        ax1.set_title(f'Custos Anuais: {processo_str}', fontweight='bold', fontsize=11, color='#1e272e')
        ax1.set_ylabel('Valor (R$)', color='#1e272e')
        ax1.tick_params(colors='#1e272e')
        for i, v in enumerate(valores):
            ax1.text(i, v + (max(valores)*0.02), f'R$ {v:,.0f}'.replace(",", "X").replace(".", ",").replace("X", "."), ha='center', fontweight='bold', color='#1e272e')

        meses = np.arange(0, 13)
        fluxo_caixa = [-investimento_total] + [-investimento_total + (ganho_total_mensal * m) for m in range(1, 13)]

        ax2.set_facecolor('#f8f9fa')
        ax2.plot(meses, fluxo_caixa, marker='o', color='#3498db', linewidth=2)
        ax2.axhline(0, color='#1e272e', linestyle='--', linewidth=1)
        ax2.fill_between(meses, 0, fluxo_caixa, where=(np.array(fluxo_caixa) >= 0), color='#2ecc71', alpha=0.3)
        ax2.fill_between(meses, 0, fluxo_caixa, where=(np.array(fluxo_caixa) < 0), color='#e74c3c', alpha=0.3)
        ax2.set_title(f'Ponto de Equilíbrio: {processo_str}', fontweight='bold', fontsize=11, color='#1e272e')
        ax2.set_xlabel('Meses', color='#1e272e')
        ax2.set_ylabel('Fluxo de Caixa Acumulado (R$)', color='#1e272e')
        ax2.set_xticks(meses)
        ax2.tick_params(colors='#1e272e')
        ax2.grid(True, linestyle=':', alpha=0.6, color='#bdc3c7')

        plt.tight_layout()
        plt.show()

botao_calcular.on_click(calcular_e_plotar)

# 4. Renderização da Interface Organizada
col_esq = widgets.VBox([html_identificacao, nome_processo, html_asis, horas_manuais, custo_hora_humano], layout=col_layout)
col_dir = widgets.VBox([html_setup, custo_squad, html_tobe, custo_nuvem_tokens, custo_mlops, ganho_estrategico], layout=col_layout)
layout_final = widgets.HBox([col_esq, col_dir], layout=widgets.Layout(width='100%', justify_content='space-between'))

display(titulo, descricao, layout_final, botao_calcular, resultado)
```
